```python
# Picking best mod, also again dropped LegalBERT

import os
import shutil

def cleanup_checkpoints(base_dir):
    """Keep only best_model and test report, delete everything else."""
    base = f'/content/drive/MyDrive/AdJUST/checkpoints/{base_dir}'
    if not os.path.exists(base):
        print(f"Skipping {base_dir} — not found")
        return
    for item in os.listdir(base):
        full_path = f'{base}/{item}'
        if item not in ('best_model', 'test_classification_report.txt'):
            if os.path.isdir(full_path):
                shutil.rmtree(full_path)
            else:
                os.remove(full_path)
    print(f"Cleaned: {base_dir}")

# Run after all models are trained
for model in ['climatebert-stage1', 'roberta-stage1', 'esgbert-stage1', 'scibert-stage1' 'finbert-stage1', 'deberta-stage1']:
    cleanup_checkpoints(model)

# Show what's left and how much space it takes
!du -sh /content/drive/MyDrive/AdJUST/checkpoints/*/best_model
```
